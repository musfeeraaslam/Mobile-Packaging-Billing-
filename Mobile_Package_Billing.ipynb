```python
def calculate_mobile_bill(package, extra_gb):
  if package.strip().lower() == "basic":
    price=1000
    free_gb = 0
  elif package.strip().lower() == "standard":
    price=1500
    free_gb = 2
  elif package.strip().lower() == "premium":
    price=1500
    free_gb = 5
  else:
    print(f"Unknown Error: {package}")
    return

  charged_gb = max(0, extra_gb - free_gb)
  extra_charges = charged_gb * 100
  final_bill = price + extra_charges

  print(f"Package: {package}")
  print(f"Package Price: {price:.0f}")
  print(f"Extra Data: {extra_gb} GB")
  print(f"Free Data: {free_gb} GB")
  print(f"Charged Data: {charged_gb} GB")
  print(f"Extra Charges: {extra_charges:.0f}")
  print(f"Final Bill: {final_bill:.0f}")

package  = input("Enter package (Basic/Standard/Premium): ")
extra_gb = int(input("Enter extra data used (GB): "))

calculate_mobile_bill(package, extra_gb)
```
